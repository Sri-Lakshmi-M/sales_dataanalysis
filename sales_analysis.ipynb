```python
# Convert 'Sale_Date' to datetime
df['Sale_Date'] = pd.to_datetime(df['Sale_Date'])
```

```python
df['Sale_Date']
```

```python
# Create a new column for Total Sales
df['Total_Sales'] = df['Quantity_Sold'] * df['Unit_Price']
```

```python
df['Total_Sales']
```

```python
df
```

```python
#Exploratory data_analysis(EDA)
#a. Sales Trends Over Time
import matplotlib.pyplot as plt
# Group by Sale_Date and sum TotalSales
daily_sales = df.groupby('Sale_Date')['Total_Sales'].sum()

# Plotting
plt.figure(figsize=(12, 6))
daily_sales.plot(kind='line', marker='o')
plt.title('Daily Sales Trend')
plt.xlabel('Order Date')
plt.ylabel('Total_Sales (£)')
plt.grid()
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
#b. Sales by Region
import seaborn as sns

# Group by Region_and_Sales_Rep and sum TotalSales
Region_and_Sales_Rep = df.groupby('Region_and_Sales_Rep')['Total_Sales'].sum().reset_index()

# Plotting
plt.figure(figsize=(8, 5))
sns.barplot(data=Region_and_Sales_Rep, x='Region_and_Sales_Rep', y='Total_Sales', palette='viridis')
plt.title('Total Sales by Region')
plt.xlabel('Region_and_Sales_Rep')
plt.ylabel('Total_Sales (£)')
plt.show()

```

```python
#sales by product
# Group by Product_Category and sum TotalSales
product_sales = df.groupby('Product_Category')['Total_Sales'].sum().reset_index()

# Plotting
plt.figure(figsize=(8, 5))
sns.barplot(data=product_sales, x='Product_Category', y='Total_Sales', palette='coolwarm')
plt.title('Total Sales by Product')
plt.xlabel('Product_Category')
plt.ylabel('Total_Sales (£)')
plt.show()
```
